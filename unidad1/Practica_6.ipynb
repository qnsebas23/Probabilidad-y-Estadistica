##**Estadistica descriptiva de los precios**

```python
precios = df['price'].dropna()
```

```python
print("Precio mínimo:", precios.min())
```

```python
print("Precio máximo:", precios.max())
```

```python
print("Moda de los precios:")
print(precios.mode())
```

```python
print("Media:", precios.mean())
```

```python
print("Mediana:", precios.median())
```

```python
print("Desviación estándar:", precios.std())
```

```python
df['price'].describe()
```

```python
print(df['price'].describe())
```

##**Histograma de precios**

```python
n = len(precios)

bins_sturges = int(np.ceil(1 + 3.322 * np.log10(n)))

print("Cantidad de datos:", n)
print("Bins según Sturges:", bins_sturges)
```

```python
plt.figure(figsize=(10,6))

plt.hist(precios, bins=bins_sturges, edgecolor='black')

plt.title('Distribución de precios de los automóviles')
plt.xlabel('Precio del automóvil')
plt.ylabel('Frecuencia')
plt.grid(axis='y', alpha=0.3)

plt.show()
```

 ## **Interpretación y conclusión**
El histograma muestra que la gran mayoría de los autos están en los rangos de precios bajos y medios. Por otro lado, hay menos vehículos costosos, lo que empuja la distribución hacia la derecha.

Los precios no están distribuidos de manera uniforme. La mayoría de los autos tienen costos bajos o moderados, y pocos tuvieron precios excesivos, así que está bien tener una distribución con cola derecha y zigzags.

##**Boxplot:** precio vs. tipo de vehículo

```python
print(df['body-style'].unique())
```

```python
plt.figure(figsize=(10,6))

sns.boxplot(data=df, x='body-style', y='price')

plt.title('Precio de los automóviles según el tipo de vehículo')
plt.xlabel('Tipo de vehículo')
plt.ylabel('Precio')
plt.xticks(rotation=30)

plt.show()
```

## **Interpretación**
 El diagrama de caja permite comparar los precios entre los diferentes tipos de vehículo. Algunas categorías tienen precios más altos y una mayor variación que otras. También se pueden observar algunos autos con precios bastante más elevados que el resto.

# **7.-Boxplot: Precio vs. ubicación del motor**

```python
plt.figure(figsize=(8,6))

sns.boxplot(data=df, x='engine-location', y='price')

plt.title('Precio de los automóviles según la ubicación del motor')
plt.xlabel('Ubicación del motor')
plt.ylabel('Precio')

plt.show()
```

El diagrama permite ver cómo se relacionan los precios con la ubicación del motor. La mayoría de los autos tiene el motor en la parte delantera, mientras que son pocos los que lo tienen en la parte trasera. Estos últimos suelen tener precios más altos que gran parte de los vehículos con motor delantero.
