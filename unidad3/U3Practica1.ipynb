```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = [0, 1, 2, 3, 4, 5]
p = [0.10, 0.25, 0.30, 0.20, 0.10, 0.05]

# 1. Verificar que las probabilidades sumen 1
suma_prob = sum(p)
print(f"Suma de probabilidades: {suma_prob:.2f}")

# 2. Elaborar la distribución de probabilidad (Gráfica)
plt.stem(x, p)
plt.title("Distribución de Probabilidad")
plt.xlabel("Número de células anormales (X)")
plt.ylabel("Probabilidad")
plt.grid(True)
plt.show()
```

```python
# 3. Valor esperado, varianza y desviación estándar
x_arr = np.array(x)
p_arr = np.array(p)

esperanza = np.sum(x_arr * p_arr)
esperanza_cuadrado = np.sum((x_arr**2) * p_arr)
varianza = esperanza_cuadrado - esperanza**2
desviacion = np.sqrt(varianza)

print(f"E(X) = {esperanza:.2f}")
print(f"E(X^2) = {esperanza_cuadrado:.2f}")
print(f"Varianza = {varianza:.2f}")
print(f"Desviación estándar = {desviacion:.2f}")

# Gráfica con media y desviación estándar
plt.stem(x, p, basefmt=" ", linefmt='b-', markerfmt='bo')
plt.title("Distribución de Probabilidad Discreta\ncon Media y Desviación Estándar")
plt.xlabel("Número de células anormales (X)")
plt.ylabel("P(X)")
plt.grid(True, linestyle='--', alpha=0.6)

# Línea vertical en la media
plt.axvline(x=esperanza, color='red', linestyle='--', label=f"Media (E[X]) = {esperanza:.2f}")

# Franja de ±1 desviación estándar
plt.axvspan(esperanza - desviacion, esperanza + desviacion, color='red', alpha=0.2, label=f"±1 Desv. Estándar (σ = {desviacion:.2f})")

plt.legend()
plt.xticks(x)
plt.show()
```

```python
# 4. Función de Distribución Acumulada
F = np.cumsum(p)

print("Función de Distribución Acumulada")
for xi, fi in zip(x, F):
    print(f"F({xi}) = {fi:.2f}")

# Gráfica de la FDA
plt.step(x, F, where='post')
plt.title("Función de Distribución Acumulada")
plt.xlabel("Número de células anormales (X)")
plt.ylabel("Probabilidad acumulada")
plt.grid(True)
plt.ylim(0, 1.05)
plt.xticks(x)
plt.show()
```

##Interpretación de resultados
Promedio esperado: el valor esperado E(X) = 2.05 indica que, en promedio, se espera encontrar aproximadamente 2 células anormales por muestra de tejido.

Variabilidad (σ = 1.28): Es moderada, la mayoría de las muestras tendrán entre 0.77 y 3.33 células.

Acumulada F(2) = 0.68: Existe un 65% de probabilidad de encontrar un máximo de 2 células anormales en una muestra al azar.
