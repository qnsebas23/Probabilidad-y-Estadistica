```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm, poisson, binom
```

```python
N_estudiantes = 500
media_peso = 70
desv_std_peso = 4
limite_inf = 68
limite_sup = 72

# probabilidad de que un estudiante pese entre 68 y 72 kg
prob_menor_72 = norm.cdf(limite_sup, media_peso, desv_std_peso)
prob_menor_68 = norm.cdf(limite_inf, media_peso, desv_std_peso)
prob_entre_68_y_72 = prob_menor_72 - prob_menor_68

# cuántos estudiantes pesan entre 68 y 72 kg
estudiantes_entre_68_y_72 = N_estudiantes * prob_entre_68_y_72

print("--- Distribución Normal (Pesos ITD) ---")
print(f"Probabilidad de estudiantes que pesan entre 68 y 72 kg: {prob_entre_68_y_72:.4f}")
print(f"Cantidad estimada de estudiantes (de 500): {round(estudiantes_entre_68_y_72)}")
```

```python
# distribución de poisson (accidentes de tránsito)
print("--- Distribución Poisson (Accidentes) ---")

# probabilidad de que no haya accidentes en 1 semana
media_accidentes = 2
prob_0_accidentes = poisson.pmf(0, media_accidentes)
print(f"1. Probabilidad de que no haya accidentes en 1 semana: {prob_0_accidentes:.6f}") #[cite: 8, 10]

# parte 2: probabilidad de que a lo sumo haya tres accidentes en 2 semanas
# la media para 2 semanas es 2 * 2 = 4
media_2_semanas = media_accidentes * 2
prob_3_o_menos = poisson.cdf(3, media_2_semanas)
print(f"2. Probabilidad de que a lo sumo haya 3 accidentes en 2 semanas: {prob_3_o_menos:.6f}") #[cite: 9, 10]
```

```python
# distribución binomial y poisson (seguros de vida)
print("--- Distribución Binomial y Poisson (Reclamaciones de Seguros) ---")

n_hombres = 5000
prob_muerte = 0.001
reclamaciones = 4

# cálculo exacto usando distribución binomial
prob_binomial_exacta = binom.pmf(reclamaciones, n_hombres, prob_muerte)
print(f"Probabilidad exacta (Binomial) de 4 reclamaciones: {prob_binomial_exacta:.6f}") #[cite: 11]

# cálculo aproximado usando distribución de poisson
# la media (mu) es n * p = 5000 * 0.001 = 5
media_muertes = n_hombres * prob_muerte
prob_poisson_aprox = poisson.pmf(reclamaciones, media_muertes)
print(f"Probabilidad aproximada (Poisson) de 4 reclamaciones: {prob_poisson_aprox:.6f}") #[cite: 11, 12, 13]
```
